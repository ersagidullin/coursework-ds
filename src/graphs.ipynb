```python
import pandas as pd
import plotly.graph_objects as go

# =========================================
# КОРРЕЛЯЦИОННАЯ МАТРИЦА
# =========================================

labels = [
    "stargazers_count",
    "subscribers_count",
    "forks_count",
    "closed_issues_count",
    "open_issues_count"
]

corr = [
    [1.00, 0.55, 0.64, 0.35, 0.28],
    [0.55, 1.00, 0.69, 0.21, 0.19],
    [0.64, 0.69, 1.00, 0.30, 0.27],
    [0.35, 0.21, 0.30, 1.00, 0.64],
    [0.28, 0.19, 0.27, 0.64, 1.00]
]

df_corr = pd.DataFrame(corr, columns=labels, index=labels)

# =========================================
# ЦВЕТОВАЯ ТЕМА
# RGB(46, 89, 240)
# =========================================

primary_blue = "rgb(46, 89, 240)"

colorscale = [
    [0.0, "#EEF2FF"],   # очень светлый
    [0.25, "#C7D2FE"],
    [0.50, "#7C9BFF"],
    [0.75, "#4F73F5"],
    [1.0, primary_blue]
]

# =========================================
# HEATMAP
# =========================================

fig = go.Figure()

fig.add_trace(
    go.Heatmap(
        z=df_corr.values,
        x=df_corr.columns,
        y=df_corr.index,

        colorscale=colorscale,

        zmin=0.19,
        zmax=1.0,

        text=df_corr.values,

        texttemplate="<b>%{text:.2f}</b>",

        textfont=dict(
            size=18,
            family="Arial",
            color="#374151"
        ),

        hovertemplate=(
            "<b>%{y}</b><br>"
            "<b>%{x}</b><br>"
            "Корреляция: %{z:.2f}"
            "<extra></extra>"
        ),

        colorbar=dict(
            thickness=35,

            tickfont=dict(
                size=16,
                color="#111827"
            ),

            outlinewidth=0
        )
    )
)

# =========================================
# СТИЛЬ
# =========================================

fig.update_layout(

    template="plotly_white",

    width=1280,
    height=900,

    paper_bgcolor="white",
    plot_bgcolor="white",

    title=dict(
        text="<b>Корреляционная матрица Спирмена</b>",
        x=0.5,

        font=dict(
            size=30,
            color="#111827",
            family="Arial"
        )
    ),

    font=dict(
        family="Arial",
        size=16,
        color="#374151"
    ),

    margin=dict(
        l=140,
        r=80,
        t=110,
        b=140
    ),

    # =====================================
    # ОСЬ X
    # =====================================

    xaxis=dict(

        side="bottom",

        tickangle=-45,

        tickfont=dict(
            size=17,
            color="#111827"
        ),

        showgrid=False,

        showline=True,
        linecolor="rgba(0,0,0,0.15)"
    ),

    # =====================================
    # ОСЬ Y
    # =====================================

    yaxis=dict(

        autorange="reversed",

        tickfont=dict(
            size=17,
            color="#111827"
        ),

        showgrid=False,

        showline=True,
        linecolor="rgba(0,0,0,0.15)"
    )
)

# =========================================
# БЕЛЫЕ РАЗДЕЛИТЕЛИ
# =========================================

fig.update_traces(
    xgap=2,
    ygap=2
)

fig.show()
```
